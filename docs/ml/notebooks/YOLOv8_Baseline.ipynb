Detection Classes:
0 - garbage
1 - potholes
2 - broken_road
3 - waterlogging
4 - infrastructure_damage(includes: damage of street light, traffic light, etc.)

```python
import os

base_path = "/content/dataset"

folders = [
    "images/train",
    "images/val",
    "labels/train",
    "labels/val"
]

for folder in folders:
    os.makedirs(os.path.join(base_path, folder), exist_ok=True)
```

```python
data_yaml = """
path: /content/dataset
train: images/train
val: images/val

names:
  0: garbage
  1: potholes
  2: broken_road
  3: waterlogging
  4: infrastructure_damage
"""

with open("/content/dataset/data.yaml", "w") as f:
    f.write(data_yaml)

print("data.yaml created successfully")
```

```python
from ultralytics import YOLO

model = YOLO("yolov8n.pt")

model.train(
    data="/content/dataset/data.yaml",
    epochs=50,
    imgsz=640,
    batch=16,
    name="baseline_yolov8"
)
```

```python
model = YOLO("runs/detect/baseline_yolov8/weights/best.pt")

model.predict(
    source="/content/dataset/images/val",
    save=True,
    name="baseline_predictions"
)
```
